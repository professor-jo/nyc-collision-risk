```python
import pandas as pd

# Read a local CSV into a DataFrame
df = pd.read_csv("Motor_Vehicle_Collisions_-_Crashes_20261119.csv")

# Inspect the dataset
print(df.shape)
# display(df.head())
```

```python

# The CSV headers are uppercase with spaces (e.g. "CRASH DATE"),
# so convert them to snake_case (e.g. "crash_date") first
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

columns = [
    "collision_id",
    "crash_date",
    "borough",
    "number_of_persons_injured",
    "number_of_persons_killed",
    "contributing_factor_vehicle_1",
    "contributing_factor_vehicle_2",
    "contributing_factor_vehicle_3",
    "contributing_factor_vehicle_4",
    "contributing_factor_vehicle_5"
]

# Keep only the fields needed for our analyses
df = df[columns]

# Create a reproducible random sample
df_sample = df.sample(
    n=min(20000, len(df)),
    random_state=42
)

df_sample.to_csv("sample.csv", index=False)
```
